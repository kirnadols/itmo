# Лабораторная работа №4
Вариант: 12

Целевая функция:
$$f(x_1,x_2) = 6x_1^2 + 5x_1x_2 + 6x_2^2 - 17x_1 - 17x_2 + 17$$

Градиент функции:
$$\nabla f(x) = \begin{pmatrix} 12x_1 + 5x_2 - 17 \\ 5x_1 + 12x_2 - 17 \end{pmatrix}$$

Начальная точка: $x^{(0)} = (-2, 5)$  
Критерий останова: $\epsilon = 0.0001$


```python
import numpy as np

epsilon = 0.0001
x0 = np.array([-2.0, 5.0])

def f(x):
    return 6*x[0]**2 + 5*x[0]*x[1] + 6*x[1]**2 - 17*x[0] - 17*x[1] + 17

def grad_f(x):
    return np.array([12*x[0] + 5*x[1] - 17, 5*x[0] + 12*x[1] - 17])
```

## 1. Метод покоординатного спуска
На каждом шаге мы фиксируем одну переменную и ищем минимум функции по другой переменной
.



```python
def bisection_search(df_1d, start_val, step=1.0, eps=0.0001):
    a = start_val
    b = start_val

    if df_1d(a) > 0:
        while df_1d(b) > 0:
            b -= step
    else:
        while df_1d(b) < 0:
            b += step

    if a > b:
        a, b = b, a


    while (b - a) > eps:
        c = (a + b) / 2.0
        if df_1d(c) == 0:
            return c
        elif df_1d(a) * df_1d(c) < 0:
            b = c
        else:
            a = c

    return (a + b) / 2.0


x = x0.copy()
iterations_cd = 0

print(f"Начальная точка: x1 = {x[0]:.5f}, x2 = {x[1]:.5f}, f(x) = {f(x):.5f}\n")

while True:
    iterations_cd += 1
    x_prev = x.copy()

    df_dx1 = lambda t: 12 * t + 5 * x[1] - 17
    x[0] = bisection_search(df_dx1, x[0], eps=epsilon)

    df_dx2 = lambda t: 5 * x[0] + 12 * t - 17
    x[1] = bisection_search(df_dx2, x[1], eps=epsilon)

    print(f"Итерация {iterations_cd}: x1 = {x[0]:.5f}, x2 = {x[1]:.5f}, f(x) = {f(x):.5f}")

    if np.linalg.norm(x - x_prev) < epsilon:
        break

print(f"\nЭкстремум найден: x1 = {x[0]:.5f}, x2 = {x[1]:.5f}")
print(f"Финальное значение f(x) = {f(x):.5f}")
print(f"Всего итераций: {iterations_cd}")
```

## 2. Метод градиентного спуска
$x^{(k+1)} = x^{(k)} - \alpha \nabla f(x^{(k)})$

```python
x = x0.copy()
alpha = 10
iterations_gd = 0

print(f"Начальная точка: x1 = {x[0]:.5f}, x2 = {x[1]:.5f}, f(x) = {f(x):.5f}\n")

while True:
    grad = grad_f(x)

    if np.linalg.norm(grad) < epsilon:
        break

    x_new = x - alpha * grad

    f_current = f(x)
    f_new = f(x_new)

    if f_new < f_current:
        x = x_new
        iterations_gd += 1
        print(f"Итерация {iterations_gd}: x1 = {x[0]:.5f}, x2 = {x[1]:.5f}, f(x) = {f_new:.5f}")
        print("ф-я уменьшается, шаг а корректен")
    else:
        print(f"Пробный шаг отклонен: f(x) = {f_new:.5f}")
        print("ф-я увеличивается, требуется изменить шаг а")

        alpha /= 2
        print(f"Новое значение шага a: {alpha:.5f}\n")

print(f"Экстремум найден: x1 = {x[0]:.5f}, x2 = {x[1]:.5f}")
print(f"Финальное значение f(x) = {f(x):.5f}")
print(f"Всего успешных итераций: {iterations_gd}")
```

## 3. Метод наискорейшего спуска
На каждом шаге направление спуска выбирается по антиградиенту, а длина шага $\alpha_k$ вычисляется путем решения задачи одномерной оптимизации:
$\min_{\alpha} f(x^{(k)} - \alpha \nabla f(x^{(k)}))$

```python
def optimize_chord(x_current, grad_current, eps=0.0001):
    def F(alpha):
        x_temp = x_current - alpha * grad_current
        grad_temp = grad_f(x_temp)
        return -np.dot(grad_current, grad_temp)

    a = 0.0
    F_a = F(a)
    b = 0.1
    F_b = F(b)

    while F_a * F_b >= 0:
        b *= 2.0
        F_b = F(b)

    while True:
        alpha_k = a - (F_a / (F_a - F_b)) * (a - b)
        F_k = F(alpha_k)

        if abs(F_k) <= eps:
            break

        if F_k > 0:
            b = alpha_k
            F_b = F_k
        else:
            a = alpha_k
            F_a = F_k

    return alpha_k

x = x0.copy()
iterations_sd = 0

print(f"Начальная точка: x1 = {x[0]:.5f}, x2 = {x[1]:.5f}, f(x) = {f(x):.5f}\n")

while True:
    grad = grad_f(x)

    if np.linalg.norm(grad) < epsilon:
        break

    iterations_sd += 1
    alpha_k = optimize_chord(x, grad, eps=epsilon)

    x = x - alpha_k * grad

    print(f"Итерация {iterations_sd}: шаг = {alpha_k:.5f}, x1 = {x[0]:.5f}, x2 = {x[1]:.5f}, f(x) = {f(x):.5f}")

print(f"\nЭкстремум найден: x1 = {x[0]:.5f}, x2 = {x[1]:.5f}")
print(f"Финальное значение f(x) = {f(x):.5f}")
print(f"Всего итераций: {iterations_sd}")
```
